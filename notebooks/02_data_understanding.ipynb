# Análise exploratória de dados

## Qual o PIB médio dos municípios por região?

```python
col_pib = 'Produto Interno Bruto, \na preços correntes\n(R$ 1.000)'
col_pib_pc = 'Produto Interno Bruto per capita, \na preços correntes\n(R$ 1,00)'
col_regiao = 'Nome da Grande Região'
col_uf = 'Sigla da Unidade da Federação'
col_municipio = 'Nome do Município'
col_ano = 'Ano'

dados.groupby(col_regiao)[[col_pib]].mean().sort_values(col_pib)
```

```python
df_pib_regiao = dados.groupby(col_regiao)[[col_pib]].mean().sort_values(col_pib)
df_pib_regiao = df_pib_regiao / 1_000_000  # R$ mil -> R$ bilhões

df_pib_regiao.plot(kind='barh', figsize=(10, 5), color='purple', legend=False)
plt.title('PIB médio dos municípios por região (2010-2023)\nFonte: IBGE | valores correntes')
plt.xlabel('PIB médio (R$ bilhões)')
plt.ylabel('Região')
plt.tight_layout()
plt.savefig('pib_medio_regiao.png', dpi=200, bbox_inches='tight')
plt.show()
```

## Quais os 10 municípios com maior PIB em 2023?

```python
top10_pib = dados.query(f"{col_ano} == 2023").sort_values(col_pib, ascending=False).head(10)
top10_pib[[col_municipio, col_uf, col_pib]]
```

```python
top10_pib = dados.query(f"{col_ano} == 2023").sort_values(col_pib, ascending=False).head(10).copy()
top10_pib[col_pib] = top10_pib[col_pib] / 1_000_000  # R$ mil -> R$ bilhões

top10_pib.plot(x=col_municipio, y=col_pib, kind='barh', figsize=(10, 5), color='green', legend=False)
plt.title('Top 10 municípios com maior PIB em 2023\nFonte: IBGE')
plt.xlabel('PIB (R$ bilhões)')
plt.ylabel('Município')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig('top10_pib_2023.png', dpi=200, bbox_inches='tight')
plt.show()
```

## Qual o percentual de municípios em cada região?

# Tratando e filtrando os dados

## Lidando com dados nulos

```python
dados.isnull().sum()
```

**Importante:** diferente do exemplo do modelo (onde os nulos são preenchidos com `0`), aqui **não vamos preencher os valores nulos com 0**. Um valor adicionado bruto igual a zero significaria "esse setor não produziu nada no município", o que seria uma informação **inventada** — não é isso que está acontecendo.

Os nulos aparecem porque o IBGE só disponibilizou o detalhamento por setor econômico até 2021; para 2022 e 2023 só existem PIB total e PIB per capita. Vamos confirmar isso abaixo.

```python
col_agro = 'Valor adicionado bruto da Agropecuária, \na preços correntes\n(R$ 1.000)'

# Em quais anos a coluna de Agropecuária está nula?
dados[dados[col_agro].isnull()][col_ano].value_counts().sort_index()
```

Confirmado: os nulos são só em 2022 e 2023, exatamente como documentado nas notas do IBGE. Por isso, vamos separar a base em dois recortes:

- **`dados` (completo, 2010–2023):** usado para analisar PIB total e PIB per capita.
- **`dados_setoriais` (2010–2021):** usado quando a análise envolver os setores econômicos (Agropecuária, Indústria, Serviços, Administração Pública).

```python
dados_setoriais = dados.query(f"{col_ano} <= 2021").copy()

# Checamos só as colunas econômicas (as que geraram os nulos em 2022/2023).
# Outras colunas da base (ex: "Região Metropolitana") têm nulos normais,
# que não têm relação com essa limitação — não fazem parte da nossa análise.
colunas_economicas = [col_pib, col_pib_pc, col_agro]
dados_setoriais[colunas_economicas].isnull().sum()
```

## Removendo registros

```python
# Verificando se existem municípios duplicados no mesmo ano (não deveria existir)
dados.duplicated(subset=[col_ano, 'Código do Município']).sum()
```

Não há registros duplicados, então não é necessário remover nenhuma linha por esse motivo. Também não há necessidade de remover colunas: diferente do exemplo do modelo (que remove a coluna `Tipo` depois de já ter sido usada), todas as colunas da nossa base seguem relevantes para as próximas fases do projeto.

## Filtros

### **1. Municípios da região Nordeste com PIB per capita acima da média nacional (2023)**

```python
media_nacional_pc = dados.query(f"{col_ano} == 2023")[col_pib_pc].mean()

selecao1 = dados[col_ano] == 2023
selecao2 = dados[col_regiao] == 'Nordeste'
selecao3 = dados[col_pib_pc] > media_nacional_pc

df_nordeste_acima_media = dados[selecao1 & selecao2 & selecao3]
df_nordeste_acima_media[[col_municipio, col_uf, col_pib_pc]].sort_values(col_pib_pc, ascending=False)
```

### **2. Municípios com PIB entre R$1 bilhão e R$5 bilhões e PIB per capita acima de R$30 mil (2023)**

```python
selecao = (
    (dados[col_ano] == 2023) &
    (dados[col_pib] >= 1_000_000) & (dados[col_pib] <= 5_000_000) &
    (dados[col_pib_pc] > 30_000)
)

df_filtro2 = dados[selecao]
df_filtro2[[col_municipio, col_uf, col_pib, col_pib_pc]].sort_values(col_pib_pc, ascending=False)
```

# Manipulando os dados

## Criando colunas numéricas

```python
# Participação (%) de cada setor no PIB do município (só faz sentido em dados_setoriais, 2010-2021)
col_industria = 'Valor adicionado bruto da Indústria,\na preços correntes\n(R$ 1.000)'
col_servicos = 'Valor adicionado bruto dos Serviços,\na preços correntes \n- exceto Administração, defesa, educação e saúde públicas e seguridade social\n(R$ 1.000)'
col_vab_total = 'Valor adicionado bruto total, \na preços correntes\n(R$ 1.000)'

dados_setoriais['Participacao_Agropecuaria_%'] = (dados_setoriais[col_agro] / dados_setoriais[col_vab_total] * 100).round(2)
dados_setoriais['Participacao_Industria_%'] = (dados_setoriais[col_industria] / dados_setoriais[col_vab_total] * 100).round(2)
dados_setoriais['Participacao_Servicos_%'] = (dados_setoriais[col_servicos] / dados_setoriais[col_vab_total] * 100).round(2)

dados_setoriais[[col_municipio, col_ano, 'Participacao_Agropecuaria_%', 'Participacao_Industria_%', 'Participacao_Servicos_%']].head()
```

```python
# Crescimento do PIB em relação ao ano anterior, por município
dados_ordenado = dados.sort_values([col_municipio, col_ano])
dados_ordenado['Crescimento_PIB_%'] = dados_ordenado.groupby('Código do Município')[col_pib].pct_change() * 100

dados_ordenado[[col_municipio, col_ano, col_pib, 'Crescimento_PIB_%']].tail(10)
```

## Criando colunas categóricas

```python
# Faixa de PIB per capita (categórica)
def faixa_pib_pc(valor):
    if valor < 15000:
        return 'Baixo'
    elif valor < 30000:
        return 'Médio'
    else:
        return 'Alto'

dados['Faixa_PIB_per_capita'] = dados[col_pib_pc].apply(faixa_pib_pc)
dados[[col_municipio, col_ano, col_pib_pc, 'Faixa_PIB_per_capita']].head()
```

```python
# Período de governo federal (categórica), útil para responder perguntas sobre evolução política x econômica
def periodo_governo(ano):
    if ano <= 2014:
        return 'Dilma 1 (2011-2014)'
    elif ano <= 2016:
        return 'Dilma 2 / Temer (2015-2016)'
    elif ano <= 2018:
        return 'Temer (2016-2018)'
    elif ano <= 2022:
        return 'Bolsonaro (2019-2022)'
    else:
        return 'Lula 3 (2023)'

dados['Periodo_Governo'] = dados[col_ano].apply(periodo_governo)
dados[[col_ano, 'Periodo_Governo']].drop_duplicates().sort_values(col_ano)
```
