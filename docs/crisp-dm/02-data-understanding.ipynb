# Tratando e filtrando os dados

## Lidando com dados nulos

```python
dados.isnull().sum()
```

**Importante:** diferente do exemplo do modelo (onde os nulos são preenchidos com `0`), aqui **não vamos preencher os valores nulos com 0**. Um valor adicionado bruto igual a zero significaria "esse setor não produziu nada no município", o que seria uma informação **inventada** — não é isso que está acontecendo.

Os nulos aparecem porque o IBGE só disponibilizou o detalhamento por setor econômico até 2021; para 2022 e 2023 só existem PIB total e PIB per capita. Vamos confirmar isso abaixo.

```python
col_agro = 'Valor adicionado bruto da Agropecuária, \na preços correntes\n(R$ 1.000)'

# Em quais anos a coluna de Agropecuária está nula?
dados[dados[col_agro].isnull()][col_ano].value_counts().sort_index()
```

Confirmado: os nulos são só em 2022 e 2023, exatamente como documentado nas notas do IBGE. Por isso, vamos separar a base em dois recortes:

- **`dados` (completo, 2010–2023):** usado para analisar PIB total e PIB per capita.
- **`dados_setoriais` (2010–2021):** usado quando a análise envolver os setores econômicos (Agropecuária, Indústria, Serviços, Administração Pública).

```python
dados_setoriais = dados.query(f"{col_ano} <= 2021").copy()

# Checamos só as colunas econômicas (as que geraram os nulos em 2022/2023).
# Outras colunas da base (ex: "Região Metropolitana") têm nulos normais,
# que não têm relação com essa limitação — não fazem parte da nossa análise.
colunas_economicas = [col_pib, col_pib_pc, col_agro]
dados_setoriais[colunas_economicas].isnull().sum()
```

# Manipulando os dados

## Criando colunas numéricas

```python
# Participação (%) de cada setor no PIB do município (só faz sentido em dados_setoriais, 2010-2021)
col_industria = 'Valor adicionado bruto da Indústria,\na preços correntes\n(R$ 1.000)'
col_servicos = 'Valor adicionado bruto dos Serviços,\na preços correntes \n- exceto Administração, defesa, educação e saúde públicas e seguridade social\n(R$ 1.000)'
col_vab_total = 'Valor adicionado bruto total, \na preços correntes\n(R$ 1.000)'

dados_setoriais['Participacao_Agropecuaria_%'] = (dados_setoriais[col_agro] / dados_setoriais[col_vab_total] * 100).round(2)
dados_setoriais['Participacao_Industria_%'] = (dados_setoriais[col_industria] / dados_setoriais[col_vab_total] * 100).round(2)
dados_setoriais['Participacao_Servicos_%'] = (dados_setoriais[col_servicos] / dados_setoriais[col_vab_total] * 100).round(2)

dados_setoriais[[col_municipio, col_ano, 'Participacao_Agropecuaria_%', 'Participacao_Industria_%', 'Participacao_Servicos_%']].head()
```

```python
# Crescimento do PIB em relação ao ano anterior, por município
dados_ordenado = dados.sort_values([col_municipio, col_ano])
dados_ordenado['Crescimento_PIB_%'] = dados_ordenado.groupby('Código do Município')[col_pib].pct_change() * 100

dados_ordenado[[col_municipio, col_ano, col_pib, 'Crescimento_PIB_%']].tail(10)
```

## Criando colunas categóricas

```python
# Faixa de PIB per capita (categórica)
def faixa_pib_pc(valor):
    if valor < 15000:
        return 'Baixo'
    elif valor < 30000:
        return 'Médio'
    else:
        return 'Alto'

dados['Faixa_PIB_per_capita'] = dados[col_pib_pc].apply(faixa_pib_pc)
dados[[col_municipio, col_ano, col_pib_pc, 'Faixa_PIB_per_capita']].head()
```

```python
# Período de governo federal (categórica), útil para responder perguntas sobre evolução política x econômica
def periodo_governo(ano):
    if ano <= 2014:
        return 'Dilma 1 (2011-2014)'
    elif ano <= 2016:
        return 'Dilma 2 / Temer (2015-2016)'
    elif ano <= 2018:
        return 'Temer (2016-2018)'
    elif ano <= 2022:
        return 'Bolsonaro (2019-2022)'
    else:
        return 'Lula 3 (2023)'

dados['Periodo_Governo'] = dados[col_ano].apply(periodo_governo)
dados[[col_ano, 'Periodo_Governo']].drop_duplicates().sort_values(col_ano)
```

# Modelagem — Separando as Entidades

A base do IBGE vem como uma única tabela "achatada" (todas as informações juntas em cada linha). Para o projeto de BI, vamos separar essas informações em **entidades** (modelo dimensional: 1 tabela fato + tabelas dimensão), que é a estrutura que o Power BI usa para funcionar bem.

**Entidades identificadas:**

- **`dim_regiao`** — as 5 grandes regiões do Brasil
- **`dim_uf`** — os 27 estados (cada um pertence a uma região)
- **`dim_municipio`** — os 5.570 municípios (cada um pertence a um estado)
- **`dim_tempo`** — os anos disponíveis (2010 a 2023)
- **`fato_pib`** — os valores de PIB e PIB per capita de cada município em cada ano (a "medida" que queremos analisar)

Essas entidades se relacionam assim: `fato_pib` → `dim_municipio` → `dim_uf` → `dim_regiao`, e `fato_pib` → `dim_tempo`. Isso é chamado de **esquema estrela** (star schema).

```python
# Nomes das colunas de código/chave que vamos usar para montar as entidades
col_cod_regiao = 'Código da Grande Região'
col_cod_uf = 'Código da Unidade da Federação'
col_sigla_uf = 'Sigla da Unidade da Federação'
col_nome_uf = 'Nome da Unidade da Federação'
col_cod_mun = 'Código do Município'
```

```python
# Dimensão Região
dim_regiao = dados[[col_cod_regiao, col_regiao]].drop_duplicates().sort_values(col_cod_regiao).reset_index(drop=True)
dim_regiao
```

```python
# Dimensão UF (Estado) — cada UF pertence a uma região (chave estrangeira: col_cod_regiao)
dim_uf = dados[[col_cod_uf, col_sigla_uf, col_nome_uf, col_cod_regiao]].drop_duplicates().sort_values(col_cod_uf).reset_index(drop=True)
dim_uf.head()
```

```python
# Dimensão Município — cada município pertence a uma UF (chave estrangeira: col_cod_uf)
dim_municipio = dados[[col_cod_mun, col_municipio, col_cod_uf]].drop_duplicates().sort_values(col_cod_mun).reset_index(drop=True)
dim_municipio.head()
```

```python
# Dimensão Tempo — os anos disponíveis na base
dim_tempo = dados[[col_ano]].drop_duplicates().sort_values(col_ano).reset_index(drop=True)
dim_tempo
```

```python
# Tabela Fato — a medida que queremos analisar (PIB e PIB per capita),
# ligada ao município (dim_municipio) e ao ano (dim_tempo)
fato_pib = dados[[col_cod_mun, col_ano, col_pib, col_pib_pc]].copy()
fato_pib.head()
```

## Verificando a modelagem

Antes de seguir, precisamos confirmar que separar os dados dessa forma não perdeu nem duplicou nenhuma informação. Fazemos isso juntando (`merge`) as tabelas de volta e comparando com a base original:

```python
# Reconstruindo os dados originais a partir das entidades separadas
reconstruido = (
    fato_pib
    .merge(dim_municipio, on=col_cod_mun)
    .merge(dim_uf, on=col_cod_uf)
    .merge(dim_regiao, on=col_cod_regiao)
)

print("Linhas na base original:", dados.shape[0])
print("Linhas depois de juntar as entidades de volta:", reconstruido.shape[0])
print("Bateu certinho?", dados.shape[0] == reconstruido.shape[0])
```

```python
print(f"Quantidade de regiões: {len(dim_regiao)}")
print(f"Quantidade de UFs: {len(dim_uf)}")
print(f"Quantidade de municípios: {len(dim_municipio)}")
print(f"Quantidade de anos: {len(dim_tempo)}")
print(f"Linhas na tabela fato: {len(fato_pib)}")
```

## Conclusão da Modelagem

Conseguimos separar a base única do IBGE em 1 tabela fato (`fato_pib`) e 4 tabelas dimensão (`dim_regiao`, `dim_uf`, `dim_municipio`, `dim_tempo`), sem perder nem duplicar nenhuma informação — o teste de reconstrução confirmou que as 77.965 linhas batem certinho.

Esse é o formato que vamos levar para a próxima fase (Preparação dos Dados / Modelagem no Power BI), já que um esquema estrela facilita a criação de medidas, relacionamentos e visualizações no dashboard.

## Tempo de acesso (consultas em pandas, na memória do Colab)

```python
import time

def medir_ms(funcao, repeticoes=5):
    tempos = []
    for _ in range(repeticoes):
        inicio = time.perf_counter()
        funcao()
        tempos.append((time.perf_counter() - inicio) * 1000)
    return round(sum(tempos) / len(tempos), 1)

print("Consulta simples (filtro por ano):", medir_ms(lambda: dados[dados[col_ano] == 2023]), "ms")
print("Consulta com JOIN (fato + município + UF):", medir_ms(lambda: fato_pib.merge(dim_municipio, on=col_cod_mun).merge(dim_uf, on=col_cod_uf)), "ms")
print("Agregação (PIB total por ano):", medir_ms(lambda: fato_pib.groupby(col_ano)[col_pib].sum()), "ms")
```
